# 電話番号・郵便番号・メールアドレスを正規表現で抜き出す

[![Open In Colab](https://colab.research.google.com/assets/colab-badge.svg)](https://colab.research.google.com/github/mamorlis/nlpcookbook/blob/main/notebooks/02_string/regex_patterns.ipynb)

NLP クックブック 02-11 ／ 文字列処理と正規化

テキストから定型的な文字列を抽出したり、個人情報としてマスクしたりしたい。

```python
import re
import unicodedata

PATTERNS = {
    "郵便番号": r"(?<![\d-])\d{3}-\d{4}(?![\d-])",
    "電話番号": r"(?<!\d)0\d{1,4}-\d{1,4}-\d{4}(?!\d)",
    "メール": r"[\w.+-]+@[\w-]+(?:\.[\w-]+)+",
    "金額": r"(?:¥|￥)?\d{1,3}(?:,\d{3})+円?|\d+円",
}

text = "〒１９１-００６５ 東京都日野市。電話 042-585-0000、携帯 090-1234-5678、連絡先 taro.yamada+nlp@example.ac.jp、参加費 3,000円"
text = unicodedata.normalize("NFKC", text)  # 先に全角を半角へ
for name, pat in PATTERNS.items():
    print(name, re.findall(pat, text))
```

```python
def mask(text):
    for name, pat in PATTERNS.items():
        if name != "金額":
            text = re.sub(pat, f"<{name}>", text)
    return text

print(mask(text))
```

- 氏名や住所は正規表現では取れない。固有表現抽出の章を参照。

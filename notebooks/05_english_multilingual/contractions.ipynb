```python
import contractions

for s in ["I'm sure you'll love it, but they don't.", "She's been there. He's happy.", "y'all gonna wanna see this", "It's John's book."]:
    print(contractions.fix(s))
```

- `'s` は is / has / 所有格の曖昧性があり、機械的な展開では区別できない。

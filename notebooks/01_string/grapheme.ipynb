`len()` だと絵文字や結合文字で文字数がずれる。人間が見る「1 文字」単位で数えたり切ったりしたい。

```python
!pip install -q regex
```

```python
import regex

text = "👨‍👩‍👧と🇯🇵でがんばる"
graphemes = regex.findall(r"\X", text)  # \X = 拡張書記素クラスタ
print(len(text), "←len()")
print(len(graphemes), "←書記素数")
print(graphemes)
print("先頭 3 文字:", "".join(graphemes[:3]))
```

- 文字数制限での切り詰めは書記素単位で行わないと、絵文字が壊れて文字化けする。
